1. # **Variables**

*Una variabile è un modo per memorizzare dati nel tuo programma. Considerala come un contenitore che contiene un valore, che può essere utilizzato e modificato nel codice.*

- *Creating Variables:*

```python
age = 13 #int
pi = 3.14 #float
name = "Alice" #string 
#per stampare il valore della variabile
print(age)
print(pi)
print(name)
#per stampare il tipo di variabile
print(type(age))
print(type(pi))
print(type(name))
```

- *Variable Naming Rules:*

I nomi delle variabili devono iniziare con una lettera o un trattino basso (_).

Possono contenere lettere, numeri e trattini bassi, ma non possono iniziare con un numero.

I nomi delle variabili sono sensibili alle maiuscole e alle minuscole (Età e età sono due parole diverse).

2. # **Data Types**

*I tipi di dati definiscono il tipo di dati che una variabile può contenere. Python ha diversi tipi di dati integrati, ma ci concentreremo su tre di base: interi, numeri in virgola mobile e stringhe.*

- ### Intergs (Int) ###

*Gli interi sono numeri interi, sia positivi che negativi, senza punto decimale.*

```python
a = 10  # Intero
b = -5 # Intero negativo
```

*I numeri interi vengono utilizzati quando è necessario contare elementi, misurare qualcosa in numeri interi o eseguire calcoli aritmetici senza frazioni.*

- ### Floats (float) ###

*I numeri in virgola mobile sono numeri con un punto decimale. Sono utilizzati per misurazioni più precise.*

```python
price = 19.99
weight = 2.5
```

*I float vengono utilizzati quando si tratta di misurazioni che richiedono precisione, come distanze, pesi o prezzi.*

- ### Strings (str) ###

*Le stringhe sono sequenze di caratteri racchiuse tra virgolette singole o doppie. Vengono utilizzate per rappresentare del testo.*

```python
greeting = "Hello, World!"
username = 'Alice'
```

*Le stringhe vengono utilizzate per memorizzare parole, frasi o qualsiasi altra informazione testuale.*

3. # **Checking Data Types** #

*È possibile verificare il tipo di una variabile utilizzando la funzione type().*

```python
age =  26
price =  9.99
name =  "Simone"
print(type(age))  # Outputs: <class 'int'>
print(type(price))  # Outputs: <class 'float'>
print(type(name))  # Outputs: <class 'str'>
```

4. # **Converting Between Data Types** #

*A volte potrebbe essere necessario convertire un tipo di dati in un altro. Questa operazione è chiamata conversione di tipo.*

```python
# Convert integer to string
age_str = str(age)

# Convert string to integer
number = int("10")

# Convert integer to float
float_number = float(5)

print(age_str)
print(type(age_str))    # Outputs: <class 'str'>
print(number)           # Outputs: <class 'int'> 
print(type(number))
print(float_number) # Outputs: <class 'float'>
print(type(float_number))
```

5. # **Simple Input and Output in Python** #



- 1.  ## **Output Using print()** ##

*La funzione print() viene utilizzata per visualizzare l'output sullo schermo. Abbiamo già visto come funziona con testo semplice, ma può essere utilizzata anche per visualizzare i risultati dei calcoli o il contenuto delle variabili.*

```python
print("Welcome to Python!")
number = 10
print("The number is", number)
```

- 2.  ## **Input Using input()** ##

*La funzione input() viene utilizzata per ricevere input dall'utente. Quando questa funzione viene chiamata, il programma si ferma e attende che l'utente digiti qualcosa e prema Invio.*

*Tutto ciò che l'utente digita viene memorizzato come stringa nel nome della variabile.*

```python
name = input("Enter your name: ")
print(name)
```

- 3. ## **Combining input() and print()** ##

*È possibile combinare input() e print() per creare programmi interattivi che rispondono all'input dell'utente.*

```python
name = input("What is your name? ")
print("Hello,", name + "!")
```

*Se si desidera lavorare con i numeri (come interi o float), è necessario convertire l'input, poiché input() restituisce sempre una stringa.*

```python
age = int(input("How old are you? "))
print("Next year, you will be", age + 1, "years old.")
```

- 4. ## **Example: Simple Calculator** ##

*Realizziamo una calcolatrice di base che chiede all'utente di inserire due numeri e poi li somma.*

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
sum_result = num1 + num2
print("The sum of", num1, "and", num2, "is", sum_result)
```
