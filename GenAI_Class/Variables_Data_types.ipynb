1. # **Variables**

*Una variabile è un modo per memorizzare dati nel tuo programma. Considerala come un contenitore che contiene un valore, che può essere utilizzato e modificato nel codice.*

- *Creating Variables:*

```python
age = 13 #int
pi = 3.14 #float
name = "Alice" #string 
#per stampare il valore della variabile
print(age)
print(pi)
print(name)
#per stampare il tipo di variabile
print(type(age))
print(type(pi))
print(type(name))
```

- *Variable Naming Rules:*

I nomi delle variabili devono iniziare con una lettera o un trattino basso (_).

Possono contenere lettere, numeri e trattini bassi, ma non possono iniziare con un numero.

I nomi delle variabili sono sensibili alle maiuscole e alle minuscole (Età e età sono due parole diverse).
