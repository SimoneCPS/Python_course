# **Set Methods in Python** #

*In Python, gli insiemi sono raccolte di elementi univoci e non ordinati. Sono utili per archiviare elementi in cui non sono ammessi duplicati e per eseguire varie operazioni come unione, intersezione e differenza. Python fornisce diversi metodi integrati per lavorare con gli insiemi.*

1. ### **add() Method** ###

*Il metodo add() aggiunge un singolo elemento all'insieme. Se l'elemento esiste già nell'insieme, non fa nulla.*

```python
#Syntax:

#set_name.add(element)

#Example:

fruits = {"apple", "banana", "cherry"}
fruits.add("orange")
print(fruits)  # Output: {'orange', 'banana', 'apple', 'cherry'}
```
