# **Set Methods in Python** #

*In Python, gli insiemi (nome_insieme = {}) sono raccolte di elementi univoci e non ordinati. Sono utili per archiviare elementi in cui non sono ammessi duplicati e per eseguire varie operazioni come unione, intersezione e differenza. Python fornisce diversi metodi integrati per lavorare con gli insiemi.*

1. ### **add() Method** ###

*Il metodo add() aggiunge un singolo elemento all'insieme. Se l'elemento esiste già nell'insieme, non fa nulla.*

```python
#Syntax:

#set_name.add(element)

#Example:

fruits = {"apple", "banana", "cherry"}
fruits.add("orange")
print(fruits)  # Output: {'orange', 'banana', 'apple', 'cherry'}
```

2. ### **update() Method** ###

*Il metodo update() aggiunge più elementi al set. È possibile passare come argomento una lista, una tupla, un set o qualsiasi iterabile.*

```python
#Syntax:

#set_name.update(iterable)

#Example:

fruits = {"apple", "banana", "cherry"}
more_fruits = ["orange", "grape"]
fruits.update(more_fruits)
print(fruits)  # Output: {'apple', 'orange', 'banana', 'grape', 'cherry'}
```

3. ### **remove() Method** ###

*Il metodo remove() rimuove un elemento specificato dall'insieme. Se l'elemento non viene trovato, genera un errore KeyError.*

```python
#Syntax:

#set_name.remove(element)

#Example:

fruits = {"apple", "banana", "cherry"}
fruits.remove("banana")
print(fruits)  # Output: {'apple', 'cherry'}
```

4. ### **discard() Method** ###

*Il metodo discard() rimuove un elemento specificato dall'insieme. Se l'elemento non viene trovato, non fa nulla (non viene generato alcun errore).*

```python
#Syntax:

#set_name.discard(element)

#Example:

fruits = {"apple", "banana", "cherry"}
fruits.discard("banana")
fruits.discard("mango")  # nessun errore anche se "mango" non è presente
print(fruits)  # Output: {'apple', 'cherry'}
```

5. ### **pop() Method** ###

*Il metodo pop() rimuove e restituisce un elemento casuale dall'insieme. Poiché gli insiemi non sono ordinati, non si sa quale elemento verrà rimosso. Se l'insieme è vuoto, viene generato un errore KeyError.*

```python
#Syntax:

#set_name.pop()

#Example:

fruits = {"apple", "banana", "cherry"}
removed_fruit = fruits.pop()
print(removed_fruit)  # Output: Random element (e.g., 'apple')
print(fruits)         # Output: Remaining elements
```

6. ### **clear() Method** ###

*Il metodo clear() rimuove tutti gli elementi dall'insieme, lasciandolo vuoto.*

```python
#Syntax:

#set_name.clear()

#Example:

fruits = {"apple", "banana", "cherry"}
fruits.clear()
print(fruits)  # Output: set()
```

7. ### **union() Method** ###

*Il metodo union() restituisce un nuovo set contenente tutti gli elementi del set originale e tutti gli elementi dei set specificati. Può accettare più set o iterabili come argomenti.*

```python
#Syntax:

#set_name.union(other_set, ...)

#Example:

set1 = {"apple", "banana"}
set2 = {"cherry", "banana"}
combined_set = set1.union(set2)
print(combined_set)  # Output: {'apple', 'cherry', 'banana'}
```

8. ### **intersection() Method** ###

*Il metodo intersection() restituisce un nuovo set contenente solo gli elementi presenti in entrambi (o in tutti) i set.*

```python
#Syntax:

#set_name.intersection(other_set, ...)

#Example:

set1 = {"apple", "banana", "cherry"}
set2 = {"banana", "cherry", "orange"}
common_set = set1.intersection(set2)
print(common_set)  # Output: {'banana', 'cherry'}
```

9. ### **difference() Method** ###

*Il metodo difference() restituisce un nuovo set contenente elementi presenti nel set originale ma non nei set specificati.*

```python
#Syntax:

#set_name.difference(other_set, ...)

#Example:

set1 = {"apple", "banana", "cherry"}
set2 = {"banana", "cherry", "orange"}
diff_set = set1.difference(set2)
print(diff_set)  # Output: {'apple'}
```

10. ### **issubset() Method** ###

*Il metodo issubset() restituisce True se tutti gli elementi del set originale sono presenti nel set specificato, altrimenti False.*

```python
#Syntax:

#set_name.issubset(other_set)

#Example:

set1 = {"apple", "banana",}
set2 = {"apple", "banana", "cherry"}
is_subset = set1.issubset(set2)
print(is_subset)  # Output: True
```

11. ### **issuperset() Method** ###

*Il metodo issuperset() restituisce True se il set originale contiene tutti gli elementi del set specificato, altrimenti False.*

```python
#Syntax:

#set_name.issuperset(other_set)

#Example:

set1 = {"apple", "banana", "cherry"}
set2 = {"apple", "banana"}
is_superset = set1.issuperset(set2)
print(is_superset)  # Output: True
```

12. ### **symmetric_difference() Method** ###

*Il metodo symmetric_difference() restituisce un nuovo set contenente elementi che si trovano in uno dei due set ma non in entrambi.*

```python
#Syntax:

#set_name.symmetric_difference(other_set)

#Example:

set1 = {"apple", "banana", "cherry"}
set2 = {"banana", "cherry", "orange"}
sym_diff_set = set1.symmetric_difference(set2)
print(sym_diff_set)  # Output: {'apple', 'orange'}
```
