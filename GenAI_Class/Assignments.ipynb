- *Assignment 4: Esercitarsi con le istruzioni print()*

*Task: Scrivere un programma Python che stampi quanto segue:*


Chiede all'utente di inserire due numeri.

Memorizza questi numeri in variabili.

Calcola e stampa la somma, la differenza, il prodotto e il quoziente dei due numeri.

Output: 

Inserisci il primo numero: 8

Inserisci il secondo numero: 2

Somma: 10

Differenza: 6

Prodotto: 16

Quoziente: 4.0

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2
print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", float(quotient))
```
