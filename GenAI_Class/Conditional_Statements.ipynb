# **Conditional Statements in Python** #

*Le istruzioni condizionali consentono al programma di prendere decisioni in base a determinate condizioni. Permettono di controllare il flusso del programma eseguendo codice diverso a seconda che una condizione sia vera o falsa.*


1. ###  **The if Statement** ###

*L'istruzione if viene utilizzata per verificare una condizione. Se la condizione è vera, viene eseguito il blocco di codice sottostante all'istruzione if.*

```python
age = 18
if age >= 18:
    print("You are an adult.") #lasciare 4 spazi di indentazione per il blocco di codice

# condizione if con input dell'utente
age = int(input("Enter your age: "))
if age >= 18:
    print("You are an adult.")    
```

2. ###  **The else Statement** ###

*L'istruzione else fornisce un'azione alternativa se la condizione if è False.*

```python
age = 16
if age >= 18:
    print("You are an adult.")
else:
    print("You are not an adult.")
```

3. ###  **The elif Statement** ###

*L'istruzione elif (abbreviazione di "else if") consente di verificare più condizioni. Se la prima condizione if è False, il programma verifica la condizione elif e così via.*

```python
age = 15
if age >= 18:
    print("You are an adult.")
elif age >= 13:
    print("You are a teenager.")
else:
    print("You are a child.")
```

4. ###  **Nesting if Statements** ###

*È anche possibile annidare le istruzioni if ​​all'interno di altre istruzioni if ​​per verificare più condizioni.*

```python
age = 18
is_student = True

if age >= 18:
    if is_student:
        print("You are an adult student.")
    else:
        print("You are an adult.")
else:
    print("You are not an adult.")
```

- Esempiio: Simple Grading System

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

## **Comparison and Logical Operators in Python** ##


1. ###  **Comparison Operators** ###

*Gli operatori di confronto vengono utilizzati per confrontare due valori. Restituiscono Vero o Falso a seconda del risultato del confronto.*

Operator	Description	Example	Result

==	Equal to	5 == 5	True

!=	Not equal to	5 != 3	True

|>	Greater than	7 > 3	True

|<	Less than	2 < 5	True

|>=	Greater than or equal to	7 >= 7	True

<=	Less than or equal to	4 <= 4	True


```python
x = 10
y = 5

print(x > y)   # True
print(x == y)  # False
print(x != y)  # True
```

2. ###  **Logical operators** ###

*Gli operatori logici vengono utilizzati per combinare più condizioni. Consentono di eseguire confronti più complessi.*

Operator	Description	Example	Result

and	| Returns True if both conditions are True	| (5 > 3) and (10 > 8) | True

or | Returns True if at least one condition is True | (5 < 3) or (10 > 8) | True

not | Reverses the result, returns False if the condition is True	not | (5 > 3) | False


```python
a = 10
b = 20
c = 5

# Using and
if a > c and b > a:
    print("Both conditions are True")

# Using or
if a > b or c < a:
    print("At least one condition is True")

# Using not
if not (a == b):
    print("a is not equal to b")
```

3. ###  **Combining Comparison and Logical Operators** ###

*È possibile combinare operatori di confronto e logici per formare condizioni complesse. Ad esempio:*

```python
age = 20
income = 30000

if age >= 18 and income >= 25000:
    print("Eligible for the loan")
else:
    print("Not eligible for the loan")
```

- Example: Checking Number Ranges

*You can use comparison and logical operators to check if a number falls within a certain range.*

```python
number = 15

if number >= 10 and number <= 20:
    print("The number is between 10 and 20")
else:
    print("The number is outside the range")
```

##  **What are Loops?** ##

*I cicli vengono utilizzati nella programmazione per eseguire ripetutamente un blocco di codice finché non viene soddisfatta una condizione specificata. Sono un concetto fondamentale nella programmazione, che consente di automatizzare attività ripetitive.*

1. ###  **while Loops** ###

*Il ciclo while ripete un blocco di codice finché una condizione specificata è vera. Il ciclo continuerà a essere eseguito finché la condizione non diventa falsa.*

```python
i = 1
while i <= 5:
    print("Iterazione", i)
    i += 1
```

2. ###  **for Loops** ###

*Il ciclo for viene utilizzato per scorrere una sequenza (ad esempio una lista, una tupla, un dizionario, un set o una stringa) o un intervallo di numeri.*

```python
#Iterating Over a List
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
#Using range() in a for Loop
for i in range(1, 6):
    print("Iteration", i)
```

```python
#utilizzo per scorrere una lista di elementi
#singolo elemento
fruits = ["apple", "banana", "cherry"]
print(fruits[0])
print(fruits[1])
print(fruits[-1])  # Last element

#iterazione con indice
i = 0
for item in fruits:
    print("Item", i, "Fruit List element;" ,item)
    i +=1
```

3. ### **. Using Loops for Repetitive Tasks** ###

*I cicli sono particolarmente utili per eseguire attività ripetitive, come l'elaborazione di ogni elemento di un elenco, la generazione di una serie di numeri o l'esecuzione di un'azione più volte.*

```python
#Sommare i numeri
total = 0
for number in range(1, 11):
    total += number
    print("Inside Loop - Adding", number, "Total so far:", total)
print("Outside Loop - The sum is:", total)
```

```python
#Conto alla rovescia con un ciclo while
countdown = 5
while countdown > 0:
    print("Counting down:", countdown)
    countdown -= 1
print("Blast off!")
```

4. ### **Breaking Out of Loops** ###

*A volte, potrebbe essere necessario uscire prima da un ciclo. È possibile farlo utilizzando l'istruzione break.*

```python
for i in range(1, 11):
    if i == 5:
        break
    print(i)

#esempio con break
fruits = ["apple", "banana", "cherry"]
i = 0
for item in fruits:
    print("Item", i, "Fruit List element;" ,item)
    i +=1
    if item == "banana":
        print ("Found banana, exiting loop.")
        break
```
