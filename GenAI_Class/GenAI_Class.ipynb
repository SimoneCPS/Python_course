**Print Statement**

*In Python, lo statement print() serve per visualizzare informazioni a schermo, tipicamente nella console o nel terminale.*

-Printing Single Items:

```python
print("Hello world!")
```

-Printing Multiple Items:

```python
# ( , = spazio)
print("Simone","Hello world!")
```

-Printing variable Items:

```python
name = "Simone"
print("My name is" , name)
```
